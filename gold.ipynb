```python
%sql
CREATE OR REPLACE TABLE tseries.tseries.gold_error_summary AS
SELECT
    machineID,
    COUNT(*) AS total_errors,
    COUNT(DISTINCT errorID) AS unique_error_types,
    MIN(datetime) AS first_error_time,
    MAX(datetime) AS last_error_time
FROM tseries.tseries.silver_errors
GROUP BY machineID;
```

```python
%sql
SELECT *
FROM tseries.tseries.gold_error_summary
ORDER BY total_errors DESC;
```

```python
%sql
CREATE OR REPLACE TABLE tseries.tseries.gold_maintenance_summary AS
SELECT
    machineID,
    COUNT(*) AS total_maintenance_events,
    COUNT(DISTINCT comp) AS components_maintained,
    MIN(datetime) AS first_maintenance_time,
    MAX(datetime) AS last_maintenance_time
FROM tseries.tseries.silver_maint
GROUP BY machineID;
```

```python
%sql
SELECT *
FROM tseries.tseries.gold_maintenance_summary
ORDER BY total_maintenance_events DESC;
```

```python
%sql
CREATE OR REPLACE TABLE tseries.tseries.gold_failure_history AS
SELECT
    machineID,
    COUNT(*) AS total_failures,
    COUNT(DISTINCT failure) AS failure_types,
    MIN(datetime) AS first_failure_time,
    MAX(datetime) AS last_failure_time
FROM tseries.tseries.silver_failures
GROUP BY machineID;
```

```python
%sql
SELECT *
FROM tseries.tseries.gold_failure_history
ORDER BY total_failures DESC;
```

predictive maintenance features

```python
%sql
CREATE OR REPLACE TABLE tseries.tseries.gold_predictive_features AS

SELECT
    m.machineID,
    m.model,
    m.age,

    /* Telemetry */
    COUNT(t.datetime) AS telemetry_records,

    AVG(t.volt) AS avg_voltage,
    MAX(t.volt) AS max_voltage,
    MIN(t.volt) AS min_voltage,

    AVG(t.rotate) AS avg_rotation,
    MAX(t.rotate) AS max_rotation,
    MIN(t.rotate) AS min_rotation,

    AVG(t.pressure) AS avg_pressure,
    MAX(t.pressure) AS max_pressure,
    MIN(t.pressure) AS min_pressure,

    AVG(t.vibration) AS avg_vibration,
    MAX(t.vibration) AS max_vibration,
    MIN(t.vibration) AS min_vibration,

    /* Errors */
    COALESCE(e.total_errors, 0) AS total_errors,
    COALESCE(e.unique_error_types, 0) AS unique_error_types,

    /* Maintenance */
    COALESCE(mt.total_maintenance_events, 0)
        AS total_maintenance_events,

    COALESCE(mt.components_maintained, 0)
        AS components_maintained,

    /* Failures */
    COALESCE(f.total_failures, 0)
        AS total_failures,

    COALESCE(f.failure_types, 0)
        AS failure_types,

    /* Target */
    CASE
        WHEN COALESCE(f.total_failures, 0) > 0
        THEN 1
        ELSE 0
    END AS failure_flag

FROM tseries.tseries.silver_machines m

LEFT JOIN tseries.tseries.silver_telemetry t
    ON m.machineID = t.machineID

LEFT JOIN tseries.tseries.gold_error_summary e
    ON m.machineID = e.machineID

LEFT JOIN tseries.tseries.gold_maintenance_summary mt
    ON m.machineID = mt.machineID

LEFT JOIN tseries.tseries.gold_failure_history f
    ON m.machineID = f.machineID

GROUP BY
    m.machineID,
    m.model,
    m.age,
    e.total_errors,
    e.unique_error_types,
    mt.total_maintenance_events,
    mt.components_maintained,
    f.total_failures,
    f.failure_types;
```

```python
%sql
SELECT *
FROM tseries.tseries.gold_predictive_features
LIMIT 20;
```

```python
%sql
SELECT
    failure_flag,
    COUNT(*) AS machine_count
FROM tseries.tseries.gold_predictive_features
GROUP BY failure_flag;
```
